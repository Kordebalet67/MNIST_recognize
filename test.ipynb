```python
import random

epochs = 50
learning_rate_max = 0.5
errors=[]
for lr in range(1, int(10*learning_rate_max)):
    for i in range (0, epochs):
        error = random.random()
        print(f'Epoch: {i + 1}, Error: {error}, learning rate: {lr/10}')
        errors.append({lr/10: error})
```

```python
import pandas as pd

df = pd.DataFrame(errors)
df
```
